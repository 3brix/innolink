## master df (two interface)

```python
import pandas as pd
import config

def prepare_ab_master(master_df):
    master_df2 = master_df.drop(columns=["interface", "fab", "vh/vl_sequence", "target", "target_residues", "target_sequence", "notes"])
    master_df_stripped = master_df2.drop_duplicates(keep="last")
    return master_df_stripped

master_df = pd.read_csv(config.MASTER_PATH, delimiter=',')
master_df_stripped = prepare_ab_master(master_df)
master_df_stripped.info()

master_df_stripped.to_csv(config.RAW_DATA_DIR / 'master_stripped.csv', index=False)

```
