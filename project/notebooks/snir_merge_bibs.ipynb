```python
import pandas as pd

snir = pd.read_csv('/scicore/home/schwede/barta0000/project/data/raw/snir/merged.csv', index_col=0)
bibs = pd.read_csv('/scicore/home/schwede/barta0000/project/data/raw/snir/260702_all_bibs_scores_w_meta_w_b2.csv', index_col=0)

bibs.head()
```

```python
bibs_clean.to_csv('/scicore/home/schwede/barta0000/project/data/raw/snir/260702_bibs_snir.csv', index=False)
```

```python
# merge snir and bibs_clean on sample column
merged = pd.merge(snir, bibs_clean, on='sample', how='inner')
merged.head()
```
