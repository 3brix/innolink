```python
# TO DO: organize (as a module), add save/load functions, add tests, add docstrings, add type hints, add logging, add CLI interface
import config
import pandas as pd


def load_and_join_predictions(org_path, ts_path=None, keep_interfaces=None, preds_path=config.PREDS_PATH,):
    """Load predictions from originals CSV and optionally target_shuffle CSV.

    Args:
        org_path: Path to originals CSV
        ts_path: Optional path to target_shuffle CSV
        keep_interfaces: Optional set of interfaces to keep. Default = {"A,B"}
        preds_path: Optional path to save the resulting CSV
    """
    # Load original dataframe
    org_df = pd.read_csv(org_path)

    # Optionally load and concatenate target_shuffle dataframe
    if ts_path is not None:
        ts_df = pd.read_csv(ts_path)
        df = pd.concat([org_df, ts_df], ignore_index=True)
    else:
        df = org_df.copy()

    # Filter interfaces
    if keep_interfaces is None:
        keep_interfaces = {"A,B"}

    df = df[df["interface"].str.strip().isin(keep_interfaces)].copy()

    # Save if output path is provided
    if preds_path:
        df.to_csv(preds_path, index=False)

    return df

# default (the ds wildcards should be specified in config...)
#df = load_and_join_predictions('/scicore/home/schwede/barta0000/mcmahon_ds/originals/predictionsv5.csv', '/scicore/home/schwede/barta0000/mcmahon_ds/target_shuffle/predictionsv4.csv')


# Use default interface "A,B" with specific output path
#df = load_and_join_predictions('/scicore/home/schwede/barta0000/mcmahon_ds/originals/predictions.csv', '/scicore/home/schwede/barta0000/mcmahon_ds/target_shuffle/predictions.csv', preds_path='/scicore/home/schwede/barta0000/project/data/raw/preds_mcmahon2.csv')

# Or use config.KEEP_INTERFACES to filter specific interfaces
df = load_and_join_predictions('/scicore/home/schwede/barta0000/alphaseq_ds/test/predictions.csv', keep_interfaces=config.KEEP_INTERFACES, preds_path=config.PREDS_PATH)

print(f"Loaded {len(df)} rows. Interfaces: {df['interface'].unique()}")
df.head()
```

```python
# Load data
df_if = pd.read_csv(config.PREDS_PATH)
df_if.info()
```

```python
# TO DO: refactor to a function (it selects numeric columns and aggregates them by mean, while taking the first value of other columns)
def aggregate_predictions(df_if):
    num_cols = df_if.select_dtypes(include=['number']).columns
    other_cols = [c for c in df_if.columns if c not in num_cols and c != 'sample']

    agg_dict = {col: 'mean' for col in num_cols}
    agg_dict.update({col: 'first' for col in other_cols})
    return agg_dict

agg_dict = aggregate_predictions(df_if)
df_1if = df_if.groupby('sample').agg(agg_dict).reset_index()

df_1if.info()
df_1if.head()
```

```python
# TO DO: refactor to a function (it drops unnecessary columns)
def drop_unnecessary_columns(df):
    return df.drop(columns=config.COLS_TO_DROP)

df_2if = drop_unnecessary_columns(df_1if)
df_2if.info()
df_2if.head()

df_2if.to_csv(config.RAW_DATA_DIR / 'predictions_2if.csv', index=False)
```
