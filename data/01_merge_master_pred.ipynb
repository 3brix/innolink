```python
# germinal mcmahon
import pandas as pd

# Load the datasets
master_df = pd.read_csv('/home/bri/pymol/data/harvey/harvey.csv', delimiter=';')
preds_df = pd.read_csv('/home/bri/pymol/data/predictions/preds_harvey_org.csv')

def create_sample_id(row):
    vhh = str(row['vhh'])
    target = str(row['target'])
    residues = str(row['target_residues'])
    
    # Check if target is HSA to apply the specific naming convention
    if target == "HSA":
        return f"{vhh}_{target}_{residues}"
    else:
        return f"{vhh}_{target}"

# 1. Create the 'sample' column in the master table
master_df['sample'] = master_df.apply(create_sample_id, axis=1)

master_df.head()
```

```python
# 2. Join the datasets on the 'sample' column
# inner join to keep only matching rows (change to 'left' if want to keep all master records)
merged_df = pd.merge(master_df, preds_df, on='sample', how='inner')

# Display the first few rows to verify
print(merged_df[['sample', 'vhh', 'target']].head())

# Save the merged result
merged_df.to_csv('merged_harvey.csv', index=False)
```
