```python
import os
from pymol import cmd
from IPython.display import Image
```

```python
output_folder = "./pymol"
os.makedirs(output_folder, exist_ok=True)
```

```python
pdb_id = "5IFJ" # 5IG7, 5IHZ, 5IJK, 5IK3"

pdb_out = os.path.join(output_folder, f"{pdb_id}.pdb")
img_out = os.path.join(output_folder, f"{pdb_id}.png")

cmd.fetch(pdb_id, type= "pdb", async_=0)
cmd.orient()

cmd.hide('all')
cmd.show('cartoon')

cmd.set('cartoon_color', 'red')
cmd.set('cartoon_fancy_helices')
cmd.set('cartoon_highlight_color', 'grey75')

cmd.bg_color('white')

cmd.set('ray_trace_mode', '1')
cmd.ray()
cmd.png(img_out, dpi=300)
```

```python
Image(img_out)
```

```python
pdb_ids = ["5IFJ", "5IG7", "5IHZ", "5IJK", "5IK3"]

for pdb_id from pdb_ids:
    cmd.reinitialize()

    pdb_out = os.path.join(output_folder, f"{pdb_id}.pdb")
    img_out = os.path.join(output_folder, f"{pdb_id}.png")
    cmd.fetch(pdb_id, async_=0)
    cmd.orient()

    cmd.hide('all')
    cmd.show('cartoon')

    cmd.set('cartoon_color', 'red')
    cmd.set('cartoon_fancy_helices')
    cmd.set('cartoon_highlight_color', 'grey75')

    cmd.bg_color('white')

    cmd.set('ray_trace_mode', '1')
    cmd.ray()
    cmd.png(img_out, dpi=300)

    Image(img_out)
```

```python
import py3Dmol

view = py3Dmol.view(width=800, height=600)

with open('./pymol/3UZE.pdb') as f:
    pdb_data = f.read()

view.addModel(pdb_data, 'pdb')

# Set style and color by chain
view.setStyle({'cartoon': {'color': 'spectrum'}})
view.setStyle({'chain': 'A'}, {'cartoon': {'color': 'blue'}})
view.setStyle({'chain': 'C'}, {'cartoon': {'color': 'orange'}})

view.zoomTo()
view.show()
```
