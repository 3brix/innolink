```python
# create table with all of them for master table
import pandas as pd

def parse_fasta_to_table(input_file, output_file):
    results = []
    
    with open(input_file, 'r') as f:
        current_id = ""
        for line in f:
            line = line.strip()
            if not line:
                continue
            
            if line.startswith(">"):
                # Clean up the header (remove '>')
                current_id = line[1:]
            else:
                # Split the sequence by the colon delimiter
                parts = line.split(':')
                
                # Ensure we have exactly 3 parts to avoid errors
                if len(parts) == 3:
                    results.append({
                        "nanobody": current_id,
                        "vhh_sequence": parts[0],
                        "vl_sequence": parts[1],
                        "target_sequence": parts[2]
                    })
                else:
                    print(f"Warning: Entry {current_id} does not have 3 parts. Skipping.")

    # Create the DataFrame and save to CSV
    df = pd.DataFrame(results)
    df.to_csv(output_file, index=False)
    print(f"Success! Table saved to {output_file}")

# Usage: Change 'input.fasta' to your actual filename
parse_fasta_to_table('mutants.fasta', '3AB_mutants_table.csv')
```
