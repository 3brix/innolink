```python
import pandas as pd
# Read complex IDs from txt file, handling potential whitespace/paths
txt_file = "/home/bri/pymol/datasets/mcmahon/complexes_list_ts.txt"

# Read the file and extract just the complex IDs (first column)
complex_ids = []
with open(txt_file, 'r') as f:
    for line in f:
        # Split by whitespace and get the first element (complex ID)
        parts = line.strip().split()
        if parts:
            complex_ids.append(parts[0])

print(f"Loaded {len(complex_ids)} complex IDs")
print(f"First few IDs: {complex_ids[:5]}")
```

```python
# Generate full paths for each complex ID
base_path = "/scicore/home/schwede/barta0000/mcmahon_ds/target_shuffle"

full_paths = []
for complex_id in complex_ids:
    path = f"{base_path}/{complex_id}/boltz_free/output/boltz_results_{complex_id}/predictions/{complex_id}/{complex_id}_model_0.cif"
    full_paths.append(path)

# Create a DataFrame with complex IDs and their paths
df = pd.DataFrame({
    'complex_id': complex_ids,
    'path': full_paths
})

print(f"Generated paths for {len(df)} complexes")
print("\nFirst few entries:")
print(df.head(10))
```

```python

# Save results to a new file or update the original
output_file = "/home/bri/pymol/datasets/mcmahon/complexes_template_ts.txt"

# Save as tab-separated file (complex_id \t path)
with open(output_file, 'w') as f:
    for complex_id, path in zip(complex_ids, full_paths):
        f.write(f"{complex_id}\t{path}\n")

print(f"✓ Saved {len(df)} entries to {output_file}")

# Also save as CSV for easier manipulation
output_csv = "/home/bri/pymol/datasets/mcmahon/complexes_template_ts_with_paths.csv"
df.to_csv(output_csv, index=False, sep='\t')
print(f"✓ Saved as CSV to {output_csv}")
```
